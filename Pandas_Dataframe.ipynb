```python
course_df = pd.DataFrame({
    'Student': ['Aman', 'Riya', 'Neha', 'Rohit', 'Priya'],
    'Course': ['Python', 'Java', 'SQL', 'Python', 'Java'],
    'Score': [85, 90, 78, 88, 95]
})

# Display column names
print(course_df.columns)

# Retrieve Student column
print(course_df['Student'])

# Create a DataFrame with Student and Score
new_course_df = course_df[['Student', 'Score']]
print(new_course_df)

# Add Batch column
course_df['Batch'] = '2026'

print(course_df)
```

```python
import numpy as np
import pandas as pd

travel_df = pd.DataFrame({
    'Place': ['Delhi', 'Agra', 'Jaipur', 'Mumbai', 'Pune']
})

# First assignment
travel_df['Code'] = np.arange(5)

print("After adding Code:")
print(travel_df)

# Create a partially indexed Series
codes = pd.Series(
    [101, 104],
    index=[1, 4]
)
travel_df['Code'] = codes

print("After replacing Code:")
print(travel_df)
```
